Q6. How do conditional statements work in Python? Illustrate with examples.

--> Conditional statement help the program to make decisions.
 Types of Conditional Statements
1. if Statement
 If something is true, do something.

Python

age = 18

if age >= 18:
    print("You can vote")
If the condition is true (age >= 18), it prints the message.

2. if...else
 If something is true, do this. Else, do that.

python

age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
3. if...elif...else
 Check more than one condition.

python

marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Keep improving!")
Python checks the conditions one by one from top to bottom.
